```python
import os

from dotenv import load_dotenv

load_dotenv()

if "GOOGLE_API_KEY" in os.environ:
    print("API Key done")
else:
    print("API Key facing some issues")
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm=ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0.4
)

print("Model Ready")
```

```python
from langchain_core.prompts import PromptTemplate

template_string="Explain the technology '{tech_term}' in exactly one short sentence for beginners"

prompt_template = PromptTemplate.from_template(template_string)

frontend_input=input("Enter tech stack name ")

formatted_prompt=prompt_template.format(tech_term=frontend_input)
```

```python
response=llm.invoke(formatted_prompt)
print(response.content)
```

```python
from langchain_core.prompts import PromptTemplate

template_string="How Compiler work in this '{tech_stack}'"

prompt_template=PromptTemplate.from_template(template_string)

input_user=input("Enter your tech stack ")
format_prompt=prompt_template.format(tech_stack=input_user)

response=llm.invoke(format_prompt)
print(response.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate

chat_template=ChatPromptTemplate.from_messages([
    ("system","you are an expert '{tech_stack}' programming language but answer my '{user_question}' in one line "),
    ("user","Here is my '{tech_stack}' and '{user_question}'")
])

data_from_db={
    "tech_stack":"python",
    "user_question":"how python actually works"
}

formatted_message=chat_template.format_messages(
    tech_stack=data_from_db["tech_stack"],
    user_question=data_from_db["user_question"]
)

response=llm.invoke(formatted_message)
print(response.content)
```
