```python
import os

from dotenv import load_dotenv

from langchain_google_genai import ChatGoogleGenerativeAI

from langchain_core.messages import HumanMessage, SystemMessage, AIMessage

load_dotenv()

if "GOOGLE_API_KEY" in os.environ:
    print("loaded")
else:
    print("not loaded")

llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", temperature=0.4)
print("Model Ready")
```

```python
print("_____ without memory test __________")

print("User: Hi my name is Abdullah")
response_1=llm.invoke("Hi my name is Abdullah")
print(f"AI: {response_1.content}")

print("User: What's my name ")
response_2=llm.invoke("Hi What's my name ? ")
print(f"AI: {response_2.content}")
```

```python
print("____with memory test_____")

chat_history=[
    SystemMessage(content="You are a helpful assistent. Keep answer short")
]

user_1="Hi my name is Abdullah & am learning genai"
print(f"User: {user_1}")

chat_history.append(HumanMessage(content=user_1))

response_3=llm.invoke(chat_history)
print(f"{response_3.content}")

chat_history.append(AIMessage(content=response_3.content))

# Turn 02

user_2="What is my name and what actually am learning ? "
print(f"User: {user_2}")
chat_history.append(AIMessage(content=user_2))
response_4= llm.invoke(chat_history)
print(f"{response_4.content}")
```
