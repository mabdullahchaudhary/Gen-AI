```python
import os

from dotenv import load_dotenv
from langchain_core.prompts import ChatPromptTemplate
from langchain_groq import ChatGroq

load_dotenv()

if "GROQ_API_KEY" in os.environ:
    print("Loaded")
else:
    print("Not Loaded")

llm=ChatGroq(
    model="llama-3.3-70b-versatile",
    temperature=0
)

print("Model Ready")
```

```python
review_template=ChatPromptTemplate.from_messages([
    ("system",""" You are a senior python developer
    Review my code provided by user.
    1) Identify any bug or bad practices.
    2) Provide the correct code.
    Keep the explanation extremely short format the beautiful code"""),
    ("user","{user_code}")
])

bad_python_code = """
def calculate_total(price, tax):
total = price + tax
    return Total
"""
```

```python
format_code=review_template.format_messages(user_code=bad_python_code)
response=llm.invoke(format_code)
print(response.content)
```
